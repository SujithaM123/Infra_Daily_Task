```python
import pandas as pd
import numpy as np

df = pd.read_csv("ecommerce.csv")

display(df)
print("Dataset shape:", df.shape)
df.info()
```

```python
print(df.head())
print("Dataset shape:", df.shape)
print("Column names:", df.columns.tolist())
df.info()
```

```python
display(df.describe(include="all"))
```

```python
print("Missing value count:")
print(df.isnull().sum())

print("\nMissing value percentage:")
print((df.isnull().sum() / len(df) * 100).round(2))
```

```python
display(df[df.isnull().any(axis=1)])
```

```python
original_df = df.copy()

df["Quantity"] = df["Quantity"].fillna(
    pd.to_numeric(df["Quantity"], errors="coerce").median()
)

df["Product"] = df["Product"].fillna("Unknown")

print(df.isnull().sum())
```

```python
df["UnitPrice"] = pd.to_numeric(df["UnitPrice"], errors="coerce")
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")

print(df.dtypes)
print("\nMissing values after conversion:")
print(df.isnull().sum())
```

```python
print("Invalid or missing prices:")
display(df[df["UnitPrice"].isna() | (df["UnitPrice"] <= 0)])

print("Invalid or negative quantities:")
display(df[df["Quantity"].isna() | (df["Quantity"] <= 0)])
```

```python
print("Exact duplicates before cleaning:", df.duplicated().sum())

df = df.drop_duplicates()

print("Exact duplicates after cleaning:", df.duplicated().sum())
print("Remaining rows:", len(df))
```

```python
print("Repeated OrderID count:", df["OrderID"].duplicated().sum())

display(df[df["OrderID"].duplicated(keep=False)].sort_values("OrderID"))
```

```python
display(df[df.duplicated(keep=False)])
```

```python
display(
    df[df["OrderID"].duplicated(keep=False)]
    [["OrderID", "Product", "City", "Quantity", "UnitPrice"]]
    .sort_values("OrderID")
)
```

```python
valid_prices = df[df["UnitPrice"].notna() & (df["UnitPrice"] > 0)]

highest_price_product = valid_prices.loc[valid_prices["UnitPrice"].idxmax()]

print("Highest-priced product:")
print("Product:", highest_price_product["Product"])
print("Unit price:", highest_price_product["UnitPrice"])
```

```python
df["Revenue"] = df["Quantity"] * df["UnitPrice"]

display(df[["OrderID", "Product", "City", "Quantity", "UnitPrice", "Revenue"]])
```

```python
df["City"] = df["City"].str.strip().str.title()

df["City"] = df["City"].replace({
    "Bengaluru": "Bangalore"
})

display(df["City"].value_counts())
```

```python
valid_sales = df[
    df["UnitPrice"].notna() &
    (df["UnitPrice"] > 0) &
    df["Quantity"].notna() &
    (df["Quantity"] > 0)
].copy()

valid_sales["Revenue"] = valid_sales["Quantity"] * valid_sales["UnitPrice"]

city_revenue = valid_sales.groupby("City")["Revenue"].sum()

display(city_revenue.sort_values(ascending=False))
```

```python
highest_revenue_city = city_revenue.idxmax()
highest_revenue = city_revenue.max()

print("Highest-revenue city:", highest_revenue_city)
print("Total revenue:", highest_revenue)
```

```python
print("Final dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nExact duplicate rows:", df.duplicated().sum())

print("\nData types:")
print(df.dtypes)

print("\nHighest-priced product:")
display(valid_prices.loc[[valid_prices["UnitPrice"].idxmax()]])

print("\nRevenue by city:")
display(city_revenue.sort_values(ascending=False))
```
